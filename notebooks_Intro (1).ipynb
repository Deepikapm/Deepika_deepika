```python
word = input("enter word : ")
shift = int(input("enter the shift size"))
encrypted_word = ""

for char in word:
    # 1. Convert letter to its Unicode number
    current_number = ord(char) - 97
    
    # 2. Add the shift value
    new_number = (current_number + shift) % 26
    
    # 3. Convert the new number back to a letter
    new_char = chr(new_number + 97)
    
    # 4. Add it to our final text
    encrypted_word = encrypted_word + new_char

print(f"Original:  {word}")   
print(f"Encrypted: {encrypted_word}") 

```

```python
word = input("enter the word")
encrypted = ""

for char in word:
    # Convert letter to 0-25 scale, add 13, wrap around with % 26, then convert back
    new_char = chr((ord(char) - 97 + 13) % 26 + 97)
    encrypted += new_char

print(f"Original:  {word}")       # Output: hello
print(f"Encrypted: {encrypted}")   # Output: uryyb

# Running the exact same code on the encrypted word brings it back!
decrypted = ""
for char in encrypted:
    new_char = chr((ord(char) - 97 + 13) % 26 + 97)
    decrypted += new_char
print(new_char)
print(f"Decrypted: {decrypted}")   # Output: hello
```
