```python
# Count uppercase, lowercase, digits, spaces, and special characters

file = open("cryptology_no.txt", "r")
contents = file.read()
file.close()

# Initialize counters
uppercase = 0
lowercase = 0
digits = 0
spaces = 0
special = 0

for ch in contents:
    if ch.isupper():
        uppercase += 1
    elif ch.islower():
        lowercase += 1
    elif ch.isdigit():
        digits += 1
    elif ch.isspace():
        spaces += 1
    else:
        special += 1

print("Uppercase letters:", uppercase)
print("Lowercase letters:", lowercase)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special characters:", special)
```

```python
# Program to accept 5 lines of text, analyze, process, and display results

#  Accept 5 lines from user and storing it  in input.txt
with open("input.txt", "w") as infile:
    print("Enter five lines related to cryptology:")
    for i in range(1, 6):
        line = input(f"Line {i}: ")
        infile.write(line + "\n")

# Step 2: Read the file and analyze
with open("input.txt", "r") as infile:
    contents = infile.read()

num_lines = contents.count("\n") + (0 if contents.endswith("\n") else 1)
num_words = len(contents.split())
num_alpha = sum(ch.isalpha() for ch in contents)
num_digits = sum(ch.isdigit() for ch in contents)
num_spaces = sum(ch.isspace() for ch in contents)
num_special = len(contents) - (num_alpha + num_digits + num_spaces)

print("\nAnalysis of input.txt:")
print("Number of lines:", num_lines)
print("Number of words:", num_words)
print("Number of alphabetic characters:", num_alpha)
print("Number of digits:", num_digits)
print("Number of spaces:", num_spaces)
print("Number of special characters:", num_special)

# Step 3: Convert text to uppercase and store in processed.txt
upper_contents = contents.upper()
with open("processed.txt", "w") as outfile:
    outfile.write(upper_contents)


with open("processed.txt", "r") as outfile:
    print("\nContents of processed.txt:\n")
    print(outfile.read())
```

```python
import re


with open("security_data.txt", "r") as f:
    contents = f.read()

# Regex pattern for IPv4 addresses
pattern = r"\b(?:\d{1,3}\.){3}\d{1,3}\b"

# Finding all matches
ip_addresses = re.findall(pattern, contents)

print("IPv4 addresses found:")
for ip in ip_addresses:
    print(ip)
```

```python
import re

#read file
with open("security_data.txt", "r") as f:
    contents = f.read()

# all  regex patterns
ip_pattern = r"\b(?:\d{1,3}\.){3}\d{1,3}\b"
email_pattern = r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}"
session_pattern = r"\bSEC-\d{4}-\d{4}\b"
hex_pattern = r"\b[0-9a-fA-F]{8,}\b"   

# matches
ips = re.findall(ip_pattern, contents)
emails = re.findall(email_pattern, contents)
sessions = re.findall(session_pattern, contents)
hexes = re.findall(hex_pattern, contents)

# storing results in pattern_results.txt
with open("pattern_results.txt", "w") as outfile:
    outfile.write("Extracted Patterns:\n\n")
    outfile.write("IPv4 Addresses:\n")
    outfile.write("\n".join(ips) + "\n\n")

    outfile.write("Email Addresses:\n")
    outfile.write("\n".join(emails) + "\n\n")

    outfile.write("Session IDs:\n")
    outfile.write("\n".join(sessions) + "\n\n")

    outfile.write("Hexadecimal Sequences:\n")
    outfile.write("\n".join(hexes) + "\n")

print("Patterns extracted and saved in pattern_results.txt")
```
